# Built-in Functions
Python comes with predefined functions called built-in functions that you can use without creating them yourself. These are very handy for common tasks in data science, like calculations, conversions, and data manipulation.
# Common Python Built-in Functions

- `len()` → returns the length of a list, string, or sequence.  
- `sum()` → adds all numbers in a list  
- `max()` → finds the largest value  
- `min()` → finds the smallest value  
- `round()` → rounds a number to a specified decimal  
- `type()` → returns the data type of a variable  
- `abs()` → gives the absolute value of a number  
- `sorted()` → returns a sorted version of a list or sequence  
- `range()` → generates a sequence of numbers  
- `print()` → displays output on the screen

```python
numbers = [1, 5, 3]
print(len(numbers))
print(max(numbers))
print(sorted(numbers))
print(sum(numbers))
print(type('abc'))
```

#User-Defined FUNCTIONS
is a function that you create yourself to perform a specific task. They help make your code organized, reusable, and easy to understand.

An example is the mean() built-in function that we can create ourselves.

```python
#Mean value
def avgval(nbs):                    #don't use mean() as its already built-in
    return sum(nbs) / len(nbs)      #return literally means just return this value
```

```python
avgval(numbers)
```

# Pandas
is a Python **library** used for data analysis and manipulation. It provides tools to easily load, organize, clean, and analyze structured data, especially in table format using structures like DataFrames.

**Series**: 1D labeled array. It is like a column of data.

```python
data = pd.Series([10, 20, 30, 40])
print(data)
```

**Dataframe**: 2-D labeled data structure, like a table with rows and columns.

```python
data = pd.DataFrame({
    "Name": ["Jimmy", "Sara", "John"],
    "Age": [25, 22, 30],
    "Score": [90, 85, 78]
})
print(data)
```

# Bar Chart

A bar chart is used to compare values between different categories. Each category is represented by a rectangular bar whose height corresponds to the value.

```python
import matplotlib.pyplot as plt   #pyplot is the SUB-library

scores = [85, 90, 88]

plt.bar(["Bill", "Sara", "John"], scores)
plt.title("Student Scores")
plt.xlabel("Students")
plt.ylabel("Scores")
plt.show()
```

WITH OUR DATAFRAME

```python
data = {
    "Day": ["Mon", "Tue", "Wed", "Thu", "Fri"],
    "Sales": [200, 250, 180, 300, 280],
    "Expenses": [150, 180, 120, 200, 190]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.plot(x="Day", y="Sales", kind="bar", title="Daily Sales")
plt.ylabel("Sales ($)")
plt.show()
```

# Line Plot

A line plot is used to show how values change over time or across an ordered sequence. Points are connected with lines so we can easily observe trends, increases, and decreases.

x=   -> selects the column used for the x-axis (horizontal axis).

y=   -> selects the column used for the y-axis (values we want to visualize).

kind="line" -> specifies that the plot type should be a line chart.

marker="o" -> adds circular markers at each data point so they are easier to see.

title=  -> adds a title to the chart.

```python
df.plot(x="Day", y=["Sales", "Expenses"], kind="line", marker="o", title="Sales vs Expenses")
plt.ylabel("Amount ($)")
plt.show()
```
