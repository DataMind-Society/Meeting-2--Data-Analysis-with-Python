We see that we have NaN (empty data) in our dataset. In Pandas, NaN stands for “Not a Number” and represents missing or undefined values in the data.

Missing data can cause problems during analysis or visualization, so we often need to handle these values before continuing our analysis.

There are two common ways to deal with missing values:

# 1) Removing Missing Values

We can remove rows that contain missing values using the **dropna()** function

```python
df_cleaned1 = df.dropna()
df_cleaned1
```

# 2) Filling Missing Values

Instead of removing data, we can replace missing values with another value using the **fillna()** function.

a) Fill with zero

```python
df_cleaned2 = df.fillna(0) #FILL WITH NUMBER INSIDE
df_cleaned2
```

Let us create a DataFrame that includes numbers as data.

b) Fill with mean value

```python
data = {
    "Day": ["Mon", "Tue", "Wed", "Thu", "Fri"],
    "Sales": [200, 250, 180, 300, 280],
    "Expenses": [150, 180, None, 200, 190]
}

df1 = pd.DataFrame(data)
df1
```

```python
df1 = df1.fillna(df1["Expenses"].mean())
df1
```
