Exercise 1: For Loop

a) Sum of all numbers in a list

```python
numbers = [3, 6, 2, 8, 4]

total = 0

for n in numbers:
    total = total + n

print(total)
```

b) Create a new list that contains **each number multiplied by 2**.

```python
numbers = [1, 2, 3, 4]

new_list = []

for n in numbers:
    new_list.append(2*n)  #.append() is a function to ADD to a list

print(new_list)
```

c) Sum of EVEN numbers between 1 and n

```python
total = 0
n = 10
for i in range(1, 11):  #range(1st, last) and increments by 1 by default. FINAL VALUE IS EXCLUSIVE.
    if i % 2 == 0:
        total = total + i

print(total)
```

# Exercise 2: Movie Ratings Analysis

You are given a dataset containing movies, their genres, and ratings.

Your goal is to explore the data, answer questions, and visualize it.

Fill in the missing parts of the code.

```python
data = {
    "Movie": ["Avengers", "Frozen", "Batman", "Inception", "Toy Story"],
    "Genre": ["Action", "Animation", "Action", "Sci-Fi", "Animation"],
    "Rating": [8.4, 7.5, 8.2, 8.8, 8.3],
    "Views": [150, 120, 130, 140, 110]
}

df = pd.DataFrame(data)
df
```

1) Display the first few rows

```python
df.head()
```

2) Display the **Rating** column

```python
df['Rating']
```

3. Find the **average** movie rating

```python
df['Rating'].mean()
```

4. What is the best rated movie?

```python
df[df['Rating'] == df['Rating'].max()]
```

5) Create a **SCATTER** plot

```python
df.plot(
    x="Views",
    y="Rating",
    kind="scatter",
    title="Views vs Rating"
)

plt.show()
```
